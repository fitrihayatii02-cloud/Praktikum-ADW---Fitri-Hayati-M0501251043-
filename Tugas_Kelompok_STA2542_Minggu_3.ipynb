2. Membuat plot ACF dari masing-masing deret hasil simulasi yang peroleh pada nomor 1 dan bandingkan. Berdasarkan plot ACF tersebut, apa yang dapat disimpulkan terkait nilai autokorelasi pada lag 1 pada model MA(1)? (Hint: Identifikasi pola nilai autokorelasi berdasarkan nilai parameter MA, kaitkan nilai autokorelasi sampel dengan nilai autokorelasi MA(1) secara teori.)

```python
from statsmodels.graphics.tsaplots import plot_acf
import matplotlib.pyplot as plt

# Membuat 6 plot dalam 1 frame (3 baris x 2 kolom)
fig, axes = plt.subplots(3, 2, figsize=(12, 12))

# Plot ACF theta = 0.1
plot_acf(ma1_theta_01, lags=10, ax=axes[0, 0])
axes[0, 0].set_title("ACF MA(1), θ = 0.1")

# Plot ACF theta = 0.3
plot_acf(ma1_theta_03, lags=10, ax=axes[0, 1])
axes[0, 1].set_title("ACF MA(1), θ = 0.3")

# Plot ACF theta = 0.7
plot_acf(ma1_theta_07, lags=10, ax=axes[1, 0])
axes[1, 0].set_title("ACF MA(1), θ = 0.7")

# Plot ACF theta = 1.0
plot_acf(ma1_theta_10, lags=10, ax=axes[1, 1])
axes[1, 1].set_title("ACF MA(1), θ = 1.0")

# Plot ACF theta = 1.5
plot_acf(ma1_theta_15, lags=10, ax=axes[2, 0])
axes[2, 0].set_title("ACF MA(1), θ = 1.5")

# Plot ACF theta = 2.0
plot_acf(ma1_theta_20, lags=10, ax=axes[2, 1])
axes[2, 1].set_title("ACF MA(1), θ = 2.0")

# Merapikan jarak antarplot
plt.tight_layout()

plt.show()
```

Berdasarkan keenam plot ACF, terlihat bahwa autokorelasi utama pada model MA(1) terdapat pada lag 1, yaitu hubungan antara suatu nilai pengamatan dengan nilai pada satu periode sebelumnya. Sementara itu, nilai autokorelasi pada lag-lag berikutnya cenderung mendekati nol dan sebagian besar berada di dalam interval kepercayaan. Hal ini sesuai dengan karakteristik model MA(1), yaitu ACF secara teori *cut off* setelah lag 1.

Berdasarkan hasil simulasi, nilai autokorelasi pada lag 1 berbeda untuk setiap nilai parameter $\theta$. Secara umum, semakin besar nilai $\theta$ dari 0.1 hingga 1.0, nilai autokorelasi pada lag 1 semakin besar secara absolut. Ketika $\theta$ lebih besar dari 1, yaitu $\theta = 1.5$ dan $\theta = 2.0$, nilai autokorelasi pada lag 1 tidak semakin besar, tetapi cenderung menurun kembali.

Secara teori, untuk model MA(1) dengan bentuk:

$$
X_t = e_t - \theta e_{t-1},
$$

nilai autokorelasi pada lag 1 adalah:

$$
\rho(1) = \frac{-\theta}{1+\theta^2}.
$$

Berdasarkan rumus tersebut, nilai autokorelasi lag 1 akan bernilai negatif karena pada simulasi digunakan bentuk $X_t = e_t - \theta e_{t-1}$. Nilai absolut autokorelasi meningkat ketika $\theta$ meningkat hingga $\theta = 1$, kemudian menurun ketika $\theta > 1$.

Nilai ACF hasil simulasi tidak sama persis dengan nilai teoritis karena nilai yang diperoleh merupakan autokorelasi sampel dari deret dengan panjang pengamatan hanya 20. Namun, secara umum pola yang diperoleh dapat dibandingkan dengan pola autokorelasi teoritis model MA(1).

4. Membuat plot ACF dari masing-masing deret hasil simulasi yang peroleh pada nomor 3 dan membandingkannya. Berdasarkan plot ACF tersebut apa yang dapat disimpulkan terkait nilai autokorelasi pada lag 1 pada model MA(1)? (Hint: Identifikasi pola interval kepercayaan dari nilai autokorelasi berdasarkan panjang observasi.

```python
from statsmodels.graphics.tsaplots import plot_acf
import matplotlib.pyplot as plt

# Membuat 4 plot dalam 1 frame (2 baris × 2 kolom)
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# ACF untuk n = 60
plot_acf(MA1_60, zero=False, ax=axes[0, 0])
axes[0, 0].set_title("ACF MA(1) dengan θ = 0.25, n = 60")

# ACF untuk n = 100
plot_acf(MA1_100, zero=False, ax=axes[0, 1])
axes[0, 1].set_title("ACF MA(1) dengan θ = 0.25, n = 100")

# ACF untuk n = 150
plot_acf(MA1_150, zero=False, ax=axes[1, 0])
axes[1, 0].set_title("ACF MA(1) dengan θ = 0.25, n = 150")

# ACF untuk n = 300
plot_acf(MA1_300, zero=False, ax=axes[1, 1])
axes[1, 1].set_title("ACF MA(1) dengan θ = 0.25, n = 300")

# Merapikan jarak antarplot
plt.tight_layout()

plt.show()
```

Berdasarkan plot ACF dari keempat deret hasil simulasi MA(1) dengan θ=0,25, terlihat bahwa autokorelasi yang paling jelas terdapat pada lag 1, sedangkan pada lag berikutnya cenderung berada di sekitar nol. Pola ini sesuai dengan karakteristik model MA(1).

Model yang digunakan adalah

$$ X_t=\varepsilon_t-\theta\varepsilon_{t-1}, $$

sehingga ACF teoritis pada lag 1 adalah

$$ \rho(1)=-\frac{\theta}{1+\theta^2}. $$

Dengan θ=0,25,

$$ \rho(1) =-\frac{0{,}25}{1+(0{,}25)^2} \approx -0{,}235. $$

Sementara itu, untuk lag \(k>1\),

$$ \rho(k)=0. $$

Dengan demikian, secara teoritis ACF MA(1) akan terputus (cut-off) setelah lag 1. Nilai ACF lag 1 pada hasil simulasi tidak harus tepat -0.235 karena merupakan ACF sampel yang dipengaruhi oleh variasi acak.

Dari keempat plot juga terlihat bahwa semakin besar panjang observasi, interval kepercayaan semakin sempit. Secara pendekatan, interval kepercayaan 95% ACF dapat dituliskan sebagai

$$ \pm\frac{1{,}96}{\sqrt{n}}. $$

Untuk \(n=60,100,150,\) dan \(300\), batas intervalnya berturut-turut sekitar

$$ \pm0{,}253,\quad \pm0{,}196,\quad \pm0{,}160,\quad \pm0{,}113. $$

Hal ini menunjukkan bahwa semakin banyak observasi, estimasi ACF menjadi semakin presisi. Oleh karena itu, pada \(n=300\) pola autokorelasi pada lag 1 lebih mudah terlihat dibandingkan \(n=60\).

<h1> Regresi Diri (Autoregressive) </h1>

Secara spesifik, proses autoregressive order p dari deret ${Y_t}$ memenuhi persamaan berikut:

<center> $Y_t=\phi_1 Y_{t-1}+\phi_2 Y_{t-2}+...+\phi_p Y_{t-p}+e_t$ </center>

Nilai $Y_t$ pada waktu sekarang merupakan kombinasi linier dari sebanyak p nilai $Y_t$ pada waktu lampau ditambah $e_t$ yang saling bebas terhadap $Y_{t-1}, Y_{t-2},...$.

<h2> AR (2) </h2>

<center> $Y_t=\phi_1 Y_{t-1}+\phi_2 Y_{t-2}+e_t$ </center>

**Syarat kestasioneran untuk deret AR (2) adalah jika dan hanya jika tiga kondisi berikut terpenuhi, yaitu:**

10\. Membuat simulasikan deret AR (2) dengan panjang pengamatan 100 dengan parameter:  
a. $\phi_1=1.5, \phi_2=-0.5$
   
   b. $\phi_1=0.75, \phi_2=-0.25$
   
   c. $\phi_1=1, \phi_2=-0.25$
   
   d. $\phi_1=0.5, \phi_2=-0.75$

```python
# AR(2) Bagian A
# psi1 = 1.5 dan psi2 = -0.5
np.random.seed(160)
psi1, psi2= 1.5,-0.5

ar2_a = [np.random.uniform(0,1),
     np.random.uniform(0,1)]
for i in range(98):
    ar2_a.append(psi1*ar2_a[i+1]+
               psi2*ar2_a[i]+
               np.random.normal(0,1))
print(ar2_a)
```

```python
# AR(2) Bagian B
# psi1 = 0.75 dan psi2 = -0.25
np.random.seed(160)
psi1, psi2= 0.75,-0.25

ar2_b = [np.random.uniform(0,1),
     np.random.uniform(0,1)]
for i in range(98):
    ar2_b.append(psi1*ar2_b[i+1]+
               psi2*ar2_b[i]+
               np.random.normal(0,1))
print(ar2_b)
```

```python
# AR(2) Bagian C
# phi1 = 1 dan phi2 = -0.25
np.random.seed(160)

psi1, psi2 = 1, -0.25

# Membuat 2 nilai awal
ar2_c = [
    np.random.uniform(0, 1),
    np.random.uniform(0, 1)
]

# Membentuk total 100 observasi
for i in range(98):
    ar2_c.append(
        psi1 * ar2_c[i+1] +
        psi2 * ar2_c[i] +
        np.random.normal(0, 1)
    )

print(ar2_c)
```

```python
# AR(2) Bagian D
# psi1 = 0.5 dan psi2 = -0.75
np.random.seed(160)

psi1, psi2 = 0.5, -0.75

# Membuat 2 nilai awal
ar2_d = [
    np.random.uniform(0, 1),
    np.random.uniform(0, 1)
]

# Membentuk total 100 observasi
for i in range(98):
    ar2_d.append(
        psi1 * ar2_d[i+1] +
        psi2 * ar2_d[i] +
        np.random.normal(0, 1)
    )

print(ar2_d)
```

```python
import matplotlib.pyplot as plt

# Membuat 4 plot dalam 1 frame (2 baris × 2 kolom)
fig, axes = plt.subplots(2, 2, figsize=(14, 8), constrained_layout=True)


# =====================
# φ₁ = 1.5 dan φ₂ = -0.5
# =====================
axes[0, 0].plot(ar2_a, '-o')
axes[0, 0].set_title(r'AR(2) dengan $\phi_1=1.5,\ \phi_2=-0.5$')
axes[0, 0].set_ylabel(r'$Y_t$')


# =====================
# φ₁ = 0.75 dan φ₂ = -0.25
# =====================
axes[0, 1].plot(ar2_b, '-o')
axes[0, 1].set_title(r'AR(2) dengan $\phi_1=0.75,\ \phi_2=-0.25$')
axes[0, 1].set_ylabel(r'$Y_t$')


# =====================
# φ₁ = 1 dan φ₂ = -0.25
# =====================
axes[1, 0].plot(ar2_c, '-o')
axes[1, 0].set_title(r'AR(2) dengan $\phi_1=1,\ \phi_2=-0.25$')
axes[1, 0].set_xlabel('Waktu')
axes[1, 0].set_ylabel(r'$Y_t$')


# =====================
# φ₁ = 0.5 dan φ₂ = -0.75
# =====================
axes[1, 1].plot(ar2_d, '-o')
axes[1, 1].set_title(r'AR(2) dengan $\phi_1=0.5,\ \phi_2=-0.75$')
axes[1, 1].set_xlabel('Waktu')
axes[1, 1].set_ylabel(r'$Y_t$')


# Menampilkan semua plot
plt.show()
```
